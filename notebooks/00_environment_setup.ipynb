# 00 — Environment Setup

Provisions the Gloucestershire Career Match pipeline:
1. Validates project paths and seed data
2. Checks core dependencies
3. Ensures the local sentence-transformers model is available

Run this notebook once before notebooks 01–05.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"
SEED_DIR = DATA_DIR / "seed"
TAXONOMY_DIR = DATA_DIR / "taxonomy"
PROCESSED_DIR = DATA_DIR / "processed"
CORPUS_DIR = DATA_DIR / "corpus"
APP_DATA_DIR = PROJECT_ROOT / "app" / "app_data"
MODEL_DIR = PROJECT_ROOT / "models" / "local_minilm_model"

for d in [PROCESSED_DIR, CORPUS_DIR, APP_DATA_DIR, APP_DATA_DIR / "faiss_index", MODEL_DIR.parent]:
    d.mkdir(parents=True, exist_ok=True)

print(f"Project root: {PROJECT_ROOT}")
```

## 1. Validate seed & taxonomy files

```python
REQUIRED = [
    SEED_DIR / "companies_seed.csv",
    SEED_DIR / "opportunities_seed.csv",
    TAXONOMY_DIR / "intake_options.yaml",
    TAXONOMY_DIR / "psych_questions.yaml",
]

missing = [p for p in REQUIRED if not p.exists()]
if missing:
    raise FileNotFoundError(f"Missing required files: {missing}")

for p in REQUIRED:
    print(f"  ✓ {p.relative_to(PROJECT_ROOT)} ({p.stat().st_size} bytes)")
```

## 3. Sentence-transformers model

Used for RAG briefings (`scripts/build_faiss_corpus.py` / notebook 05). Notebook 02 embeddings are parked research and not required for live matching.

```python
from sentence_transformers import SentenceTransformer

if (MODEL_DIR / "config.json").exists():
    model = SentenceTransformer(str(MODEL_DIR))
    print(f"Loaded local model from {MODEL_DIR}")
else:
    print("Downloading all-MiniLM-L6-v2...")
    model = SentenceTransformer("all-MiniLM-L6-v2")
    model.save(str(MODEL_DIR))
    print(f"Saved to {MODEL_DIR}")

vec = model.encode(["Gloucestershire cyber apprenticeship"])
print(f"Embedding dim: {vec.shape[1]}")
print("Model verification passed.")
```
